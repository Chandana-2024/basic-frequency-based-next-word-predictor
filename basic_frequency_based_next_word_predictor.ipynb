```python
import random
from collections import defaultdict, Counter

sentences = [
    "i love learning ai",
    "i love learning python",
    "i love building projects",
    "i enjoy learning python",
    "i enjoy building ai projects",
    "i love building ai",
    "i enjoy learning ai",
]

# Learn which word follows each previous word
next_word_counts = defaultdict(Counter)

for sentence in sentences:
    words = sentence.split()
    words = ["<START>"] + words + ["<END>"]

    for i in range(len(words) - 1):
        current_word = words[i]
        next_word = words[i + 1]

        next_word_counts[current_word][next_word] += 1

print("Training complete!")
print("Contexts learned:", len(next_word_counts))
```

```python
def get_probabilities(context):
    counts = next_word_counts.get(context, Counter())
    total = sum(counts.values())

    if total == 0:
        return {}

    return {
        word: count / total
        for word, count in counts.items()
    }


context = "Building"
probabilities = get_probabilities(context)

for word, prob in sorted(
    probabilities.items(),
    key=lambda item: item[1],
    reverse=True
):
    print(f"{word:10} {prob:.2%}")
```

```python
def greedy_decode(probabilities):
    if not probabilities:
        return "<NONE>"
    return max(probabilities, key=probabilities.get)


def sample_decode(probabilities):
    if not probabilities:
        return "<NONE>"
    words = list(probabilities.keys())
    probs = list(probabilities.values())

    return random.choices(
        words,
        weights=probs,
        k=1
    )[0]


print("Greedy:", greedy_decode(probabilities))

print("Sampling:")
for _ in range(10):
    print(sample_decode(probabilities))
```
